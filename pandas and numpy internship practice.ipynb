```python
import pandas as pd

data = {
    "Name": ["Yash", "Babu", "Keerthi", "Rinthiya", "Joel"],
    "Python": [85, 70, 90, 45, 65],
    "Java": [80, 75, 88, 50, 70],
    "SQL": [90, 65, 95, 55, 75],
    "ML": [88, 72, 92, 40, 68]
}

df = pd.DataFrame(data)

subjects = ["Python", "Java", "SQL", "ML"]
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)

print("Highest:", df["Total"].max())
print("Lowest:", df["Total"].min())

df["Result"] = df[subjects].apply(
    lambda x: "Pass" if (x >= 50).all() else "Fail", axis=1
)

df["Grade"] = df["Average"].apply(
    lambda x: "A" if x >= 80 else "B" if x >= 60 else "C" if x >= 50 else "F"
)

print("Top 3:")
print(df.nlargest(3, "Total"))

print("Bottom 3:")
print(df.nsmallest(3, "Total"))
print("Subject Average:\n", df[subjects].mean())
print("Subject Highest:\n", df[subjects].max())
print("Subject Lowest:\n", df[subjects].min())

print("Pass %:", (df["Result"] == "Pass").mean() * 100)

df = df.sort_values("Total", ascending=False)

df.to_csv("student_performance.csv", index=False)

print(df)
```

```python
import pandas as pd

data = {
    "Order_ID": range(1, 51),
    "Date": pd.date_range("2026-01-01", periods=50),
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard", "Monitor"] * 10,
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Electronics"] * 10,
    "City": ["Chennai", "Bangalore", "Mumbai", "Delhi", "Hyderabad"] * 10,
    "Quantity": [10, 20, 50, 30, 15] * 10,
    "Price": [50000, 25000, 500, 1500, 12000] * 10
}

df = pd.DataFrame(data)


df["Revenue"] = df["Quantity"] * df["Price"]


print("Total Revenue:", df["Revenue"].sum())
print("Average Order:", df["Revenue"].mean())
print("Highest Order:\n", df.loc[df["Revenue"].idxmax()])
print("Lowest Order:\n", df.loc[df["Revenue"].idxmin()])


print("City Revenue:\n", df.groupby("City")["Revenue"].sum())
print("Category Revenue:\n", df.groupby("Category")["Revenue"].sum())
print("Product Revenue:\n", df.groupby("Product")["Revenue"].sum())


print("Best Selling Product:",
      df.groupby("Product")["Quantity"].sum().idxmax())


print("Highest Revenue City:",
      df.groupby("City")["Revenue"].sum().idxmax())


print("Top 5 Orders:\n", df.nlargest(5, "Revenue"))


print("Products Quantity > 100:\n",
      df.groupby("Product")["Quantity"].sum()
        [lambda x: x > 100])


df = df.sort_values("Revenue", ascending=False)


df.to_csv("sales_report.csv", index=False)

print(df)
```

```python
import numpy as np 
import pandas as pd 

data = {
    "Name": ["Yash", "Babu", "Keerthi", "Joel", "Yash"],
    "Course": ["Python", "Java", "Python", "Java", "Python"],
    "Python": [85, 70, 90, 65, 85],
    "Java": [80, 75, 88, 60, 80],
    "SQL": [90, 68, 95, np.nan, 90]
}

df = pd.DataFrame(data)

New_Student={
    "Name":"Ravi",
    "Course":"Python",
    "Python": 75,
    "Java": 88,
    "SQL": 90

}
df.loc[len(df)] = New_Student
df

subjects = ["Python","Java","SQL"]
df["Total"]=np.sum(df[subjects], axis = 1)


df["Average"] = np.mean(df[subjects], axis = 1)


df["Grade"] = np.where(df["Average"]>80,"A", 
                       np.where (df["Average"]>60,"B",
                       np.where(df["Average"]> 50,"C","D")  
                                       ))

df


df["Result"] = np.where(
                       (df[subjects]>=50).all (axis =1),"Pass","Fail")

print("\n Top student:")
print(df.loc [df["Total"].idxmax()])

print("Bottom Student:")
print(df.loc[df["Total"].idxmin()])

print("\n Search name:")
print(df[df["Name"]=="Yash"])

print("\n Python_students:")
print(df[df["Course"] == "Python"])

df = df.sort_values('Total',ascending=False)
df

df = df.drop_duplicates() 

df["SQL"]=df["SQL"].fillna(df["SQL"].mean())
df




```

```python
import numpy as np
import pandas as pd

data = {
    "Order_ID": range(1, 11),
    "Product": ["Rice", "Oil", "Milk", "Soap", "Rice",
                "Oil", "Bread", "Milk", "Soap", "Bread"],
    "Category": ["Grocery", "Grocery", "Dairy", "Personal",
                 "Grocery", "Grocery", "Bakery", "Dairy",
                 "Personal", "Bakery"],
    "City": ["Chennai", "Delhi", "Chennai", "Mumbai", "Delhi",
             "Chennai", "Mumbai", "Delhi", "Chennai", "Mumbai"],
    "Quantity": [10, 5, 20, 15, 8, 6, 12, 18, 10, 14],
    "Price": [1200, 1800, 60, 50, 1200, 1800, 40, 60, 50, 40]
}

df = pd.DataFrame(data)


print(df)


print("Total Quantity:", df["Quantity"].sum())
print("Average Price:", df["Price"].mean())

df["Revenue"] = df["Quantity"] * df["Price"]

print("Total Revenue:", df["Revenue"].sum())
print("Average Revenue:", df["Revenue"].mean())

print("Highest Sale:")
print(df.loc[df["Revenue"].idxmax()])

print("Lowest Sale:")
print(df.loc[df["Revenue"].idxmin()])


best = df.groupby("Product")["Quantity"].sum().idxmax()
print("Best Selling Product:", best)


print(df.groupby("Category")["Revenue"].sum())


print(df.groupby("City")["Revenue"].sum())


product_revenue = df.groupby("Product")["Revenue"].sum()
print(product_revenue)

print(product_revenue.nlargest(5))

print(df.nlargest(10, "Revenue"))

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())


df = df.drop_duplicates()


df = df.sort_values("Revenue", ascending=False)


df.to_csv("supermarket_sales_report.csv", index=False)


print("\nFinal Sales Summary")
print("Total Revenue:", df["Revenue"].sum())
print("Average Revenue:", df["Revenue"].mean())
print("Best Product:", df.groupby("Product")["Quantity"].sum().idxmax())
print("Highest Revenue City:", df.groupby("City")["Revenue"].sum().idxmax())

print("\nFinal Report:")
print(df)
```
